````python
import numpy as np

R = 8.314  # J / (mol K)


# -----------------------------
# Equation of State
# -----------------------------

def ideal_gas_pressure(T, V, n=1.0):
    """
    Ideal gas equation of state:
    P = nRT / V
    """
    return n * R * T / V


# -----------------------------
# Thermodynamic derivatives
# -----------------------------

def dP_dT_at_constant_V(T, V, n=1.0):
    """
    Partial derivative (∂P/∂T)_V for an ideal gas
    """
    return n * R / V


def dP_dV_at_constant_T(T, V, n=1.0):
    """
    Partial derivative (∂P/∂V)_T for an ideal gas
    """
    return -n * R * T / V**2
````

````python
# Define S and P from F
S = -sp.diff(F, T)  # S = - (∂F/∂T)_V
P = -sp.diff(F, V)  # P = - (∂F/∂V)_T

# Maxwell relation from F:
# (∂S/∂V)_T ?= (∂P/∂T)_V
lhs = sp.diff(S, V)  # (∂S/∂V)_T
rhs = sp.diff(P, T)  # (∂P/∂T)_V

sp.simplify(S), sp.simplify(P), sp.simplify(lhs), sp.simplify(rhs), sp.simplify(lhs - rhs)
````

````python
import numpy as np

# numeric constants
n_val = 1.0
R_val = 8.314
T0 = 300.0
V0 = 0.025

def F_numeric(T, V, n=n_val, R=R_val):
    # choose g(T)=0 for simplicity; doesn't affect mixed partials
    return -n*R*T*np.log(V)

# central difference mixed partial:
# ∂²F/∂T∂V ≈ [F(T+h, V+k) - F(T+h, V-k) - F(T-h, V+k) + F(T-h, V-k)] / (4hk)
h = 1e-4  # step in T
k = 1e-6  # step in V

d2F_num = (
    F_numeric(T0 + h, V0 + k) - F_numeric(T0 + h, V0 - k)
    - F_numeric(T0 - h, V0 + k) + F_numeric(T0 - h, V0 - k)
) / (4*h*k)

# analytic mixed partial: ∂²F/∂T∂V = -nR/V
d2F_analytic = -n_val*R_val/V0

d2F_num, d2F_analytic, d2F_num - d2F_analytic
````

````python
def dF_dT_numeric(T, V, h=1e-4):
    return (F_numeric(T + h, V) - F_numeric(T - h, V)) / (2*h)

def dF_dV_numeric(T, V, k=1e-6):
    return (F_numeric(T, V + k) - F_numeric(T, V - k)) / (2*k)

def S_numeric(T, V):
    return -dF_dT_numeric(T, V)

def P_numeric(T, V):
    return -dF_dV_numeric(T, V)

def dS_dV_numeric(T, V, k=1e-6):
    return (S_numeric(T, V + k) - S_numeric(T, V - k)) / (2*k)

def dP_dT_numeric(T, V, h=1e-4):
    return (P_numeric(T + h, V) - P_numeric(T - h, V)) / (2*h)

lhs_num = dS_dV_numeric(T0, V0)
rhs_num = dP_dT_numeric(T0, V0)

lhs_analytic = n_val*R_val/V0
rhs_analytic = n_val*R_val/V0

(lhs_num, rhs_num, lhs_num - rhs_num), (lhs_analytic, rhs_analytic)
````

````python
def dF_dT_numeric(T, V, h=1e-4):
    return (F_numeric(T + h, V) - F_numeric(T - h, V)) / (2*h)

def dF_dV_numeric(T, V, k=1e-6):
    return (F_numeric(T, V + k) - F_numeric(T, V - k)) / (2*k)

def S_numeric(T, V):
    return -dF_dT_numeric(T, V)

def P_numeric(T, V):
    return -dF_dV_numeric(T, V)

def dS_dV_numeric(T, V, k=1e-6):
    return (S_numeric(T, V + k) - S_numeric(T, V - k)) / (2*k)

def dP_dT_numeric(T, V, h=1e-4):
    return (P_numeric(T + h, V) - P_numeric(T - h, V)) / (2*h)

lhs_num = dS_dV_numeric(T0, V0)
rhs_num = dP_dT_numeric(T0, V0)

lhs_analytic = n_val*R_val/V0
rhs_analytic = n_val*R_val/V0

(lhs_num, rhs_num, lhs_num - rhs_num), (lhs_analytic, rhs_analytic)
````

````python
def delta_S_isochoric(T1, T2, V, n=1.0, Cv=20.785, N=20000):
    # Cv ~ 5/2 R for monatomic ideal gas (default)
    Ts = np.linspace(T1, T2, N)
    integrand = n*Cv/Ts
    return np.trapezoid(integrand, Ts)

def delta_S_isothermal(V1, V2, T, n=1.0, R=8.314, N=20000):
    Vs = np.linspace(V1, V2, N)
    integrand = n*R/Vs
    return np.trapezoid(integrand, Vs)
````
